# 01 — Feature Engineering & Baseline LSTM

This notebook is the entry point of the pipeline. It does two things:

1. **Uses the engineered feature set** already present in `data/processed/NVDA_features.csv`
   (`date, open, high, low, close, volume` plus the indicator columns). If any indicator column
   is missing, it's computed from the raw OHLCV columns and **appended back into that same
   CSV** — this notebook never writes a separate `features.csv`.
2. **Trains a single baseline LSTM** on the *whole* dataset (no regime splitting) to predict
   next-day return, and plots **actual vs predicted price** over time.

This baseline matters for two reasons: it's the simplest possible model to sanity-check the
feature set and training loop against, and it's the fallback model used later whenever a
regime-specific model isn't available (too little history for that regime, etc.).

**Output of this notebook:** `models/baseline_lstm.pt` (trained weights) and
`reports/baseline_results.pkl` (training history + test predictions), both consumed by the
next two notebooks. All figures are saved into `reports/`.

## 2. Indicator calculation

> **Before writing any code — run the cell below as-is first.**

> - **If your dataset is already complete** (all 39 columns present), the cell will print `"Dataset complete — skipping indicator calculation."` and exit early. You do not need to implement anything.

The six indicator groups you may need to implement (only if flagged as missing):

1. **RSI (14-period)** — 14-period Relative Strength Index scaled to **0–1**.  
   Use rolling average gains and losses from `close.diff()`, then apply the classic RS formula.

2. **MACD family** — `macd` (EMA12 − EMA26), `macd_signal` (9-period EMA of MACD), `macd_hist` (their difference).  
   `ema_12` and `ema_26` are already computed above each TODO block.

3. **Bollinger Bands** — `bb_upper`, `bb_lower`, `bb_width` (band width / SMA), `bb_pct_b` (price position 0–1).  
   Use `sma_20` (already computed) and a 20-day rolling std.

4. **Stochastic Oscillator (14, 3)** — `stoch_k` from the 14-period high/low range, `stoch_d` as its 3-period rolling mean. Scale **both to 0–1**.

5. **On-Balance Volume** — `obv` (cumulative signed volume), `obv_change` (percentage change of OBV).

6. **VWAP & ratio** — Cumulative `vwap`, and `vwap_ratio = close / vwap − 1`.


```python
# ── Step 1: detect which required columns are already present ────────────────
missing_cols = [col for col in FINAL_HEADER if col not in raw.columns]

if not missing_cols:
    print("Dataset complete — skipping indicator calculation.")
    print(f"All {len(FINAL_HEADER)} required columns found. Loading directly.")
    featured = raw[FINAL_HEADER].copy()
else:
    print(f"{len(missing_cols)} column(s) missing from your dataset: {missing_cols}")
    print("Implement the TODO blocks below that correspond to the missing columns, then re-run.\n")

    def calculate_indicators(df: pd.DataFrame) -> pd.DataFrame:
        """Compute only the missing indicator columns and merge them back."""
        df = df.copy().sort_values("date").reset_index(drop=True)
        close, high, low, volume = df["close"], df["high"], df["low"], df["volume"]

        # --- Returns (already implemented) ---
        df["return_1"]     = close.pct_change(1)
        df["return_5"]     = close.pct_change(5)
        df["log_return_1"] = np.log(close / close.shift(1))

        # --- Moving averages & trend ratios (already implemented) ---
        df["sma_20"]  = close.rolling(20).mean()
        df["sma_50"]  = close.rolling(50).mean()
        df["sma_200"] = close.rolling(200).mean()
        df["sma_ratio_20"] = close / df["sma_20"] - 1
        df["sma_ratio_50"] = close / df["sma_50"] - 1

        df["ema_12"] = close.ewm(span=12, adjust=False).mean()
        df["ema_20"] = close.ewm(span=20, adjust=False).mean()
        df["ema_26"] = close.ewm(span=26, adjust=False).mean()
        df["ema_ratio_20"] = close / df["ema_20"] - 1

        # --- TODO 1: RSI(14), scaled to 0-1 ---
        # Only implement if "rsi_14" is in missing_cols.
        # Compute a 14-period RSI using rolling average gains and losses.
        # Scale the result to the range [0, 1] (divide by 100).
        # TODO: implement df["rsi_14"] here
        delta=close.diff()
        gain=delta.clip(lower=0) #clip(lower=0)-keep +ve values , replace -ve values with 0
        loss=delta.clip(upper=0)
        avg_gain=gain.rolling(14).mean()
        avg_loss=loss.rolling(14).mean()
        rs=avg_gain/avg_loss
        df['rsi_14']=(100-(100/(1+rs)))/100



        # --- TODO 2: MACD family ---
        # Only implement if "macd", "macd_signal", or "macd_hist" are in missing_cols.
        # ema_12 and ema_26 are already computed above.
        # TODO: implement df["macd"], df["macd_signal"], df["macd_hist"] here
        df['macd']=df['ema_12']-df['ema_26']
        df['macd_signal']=df['macd'].ewm(span=9,adjust=False).mean()
        df['macd_hist']=df['macd']-df['macd_signal'] #>0-short term trend is more than long term and vice versa

        # --- TODO 3: Bollinger Bands (20-day, 2-sigma) ---
        # Only implement if "bb_upper", "bb_lower", "bb_width", or "bb_pct_b" are in missing_cols.
        # TODO: implement df["bb_upper"], df["bb_lower"], df["bb_width"], df["bb_pct_b"] here
        roll_std_20=close.rolling(20).std()
        df['bb_upper']=df['sma_20']+2*roll_std_20
        df['bb_lower']=df['sma_20']-2*roll_std_20
        df['bb_width']=(df['bb_upper']-df['bb_lower'])/df['sma_20'] #measures volatility
        df['bb_pct_b']=(close-df['bb_lower'])/(df['bb_upper']-df['bb_lower'])

        # --- Rate of change (already implemented) ---
        df["roc_5"]  = close.pct_change(5)
        df["roc_10"] = close.pct_change(10)

        # --- ATR(14) (already implemented) ---
        prev_close = close.shift(1)
        tr = pd.concat([
            high - low,
            (high - prev_close).abs(),
            (low  - prev_close).abs(),
        ], axis=1).max(axis=1) # .max(axis=1) takes maximum across each row
        df["atr_14"]    = tr.rolling(14).mean() #measures avg recent volatility
        df["atr_ratio"] = df["atr_14"] / close #normalizes ATR

        # --- TODO 4: Stochastic Oscillator (14, 3), scaled to 0-1 ---
        # Only implement if "stoch_k" or "stoch_d" are in missing_cols.
        # TODO: implement df["stoch_k"] and df["stoch_d"] here
        low_14=low.rolling(14).min()
        high_14=high.rolling(14).max()
        df['stoch_k']=(close-low_14)/(high_14-low_14) #(close to 1):strong recent buying pressure,close to 0: recent selling pressure
        df['stoch_d']=df['stoch_k'].rolling(3).mean()


        # --- TODO 5: On-Balance Volume ---
        # Only implement if "obv" or "obv_change" are in missing_cols.
        # TODO: implement df["obv"] and df["obv_change"] here
        direction=np.sign(close.diff()).fillna(0)
        df['obv']=(direction*volume).cumsum() #obv keeps on increasing -strong buying pressure ,decreasing-strong selling pressure
        df['obv_change']=df['obv'].pct_change()


        # --- TODO 6: VWAP (cumulative) and ratio ---
        # Only implement if "vwap" or "vwap_ratio" are in missing_cols.
        # TODO: implement df["vwap"] and df["vwap_ratio"] here
        df['vwap']=(close*volume).cumsum()/volume.cumsum()
        df['vwap_ratio']=close/df['vwap']-1  # +ve vwap- buying pressure and -ve vwap-selling pressure

        # --- Z-score of price vs 20-day mean (already implemented) ---
        roll_std = close.rolling(20).std()
        df["z_score_20"] = (close - df["sma_20"]) / roll_std #today's price is 'z_score_20' standard deviation above/below its avg

        # --- Targets (already implemented) ---
        df["future_return"]    = close.shift(-1) / close - 1 #computes tomorrow's return
        df["target_direction"] = (df["future_return"] > 0).astype(int) #1-stock will go up tomorrow, =-stock will not go up tomorrow

        return df

    featured = calculate_indicators(raw[REQUIRED_OHLCV])
    featured = featured[FINAL_HEADER]
    featured.to_csv(DATA_PATH, index=False)
    print(f"Missing columns computed and saved to {DATA_PATH}")

featured = featured.dropna().reset_index(drop=True)
print(f"Final feature set: {len(featured)} rows, {len(featured.columns)} columns")
featured.head()
```

## 3. Feature columns, scaling, and sequence construction

The model never sees raw prices directly — only the ratio/percentage-style derived features in `FEATURE_COLUMNS`.

**Your tasks:**

1. **`make_sequences(features, targets, seq_len)`** — Given a 2-D array of features and a 1-D array of targets, build overlapping windows.  
   For each index `i` from `seq_len` to `len(features)`, the input window is `features[i - seq_len : i]` and the label is `targets[i]`.  
   Return two NumPy arrays of dtype `float32`: `X` with shape `(N, seq_len, n_features)` and `y` with shape `(N,)`.

2. **`winsorise(arr, n_std=3.0)`** — Clip extreme values in a 1-D return array.  
   Compute a symmetric range of `±(n_std × arr.std())` and clip values outside it. Return the clipped array.

The scaler fitting, splitting, and sequence-building calls below are already written — you only need to implement the two functions above.


```python
print(f"{len(FEATURE_COLUMNS)} input features:", FEATURE_COLUMNS)


def make_sequences(features: np.ndarray, targets: np.ndarray, seq_len: int): #ndarray-N dimensional array,converts data stored from tables into sliding window of prev days
    """
    Build (X, y) pairs from a scaled feature array and a target array.

    For each position i in [seq_len, len(features)), produce:
      X[i] = features[i - seq_len : i]   shape: (seq_len, n_features)
      y[i] = targets[i]

    Returns two float32 NumPy arrays: X of shape (N, seq_len, n_features)
    and y of shape (N,).
    """
    # TODO: build xs and ys lists by iterating over valid indices,
    xs,ys=[],[] #stores all input and output
    for i in range(seq_len,len(features)): #let seq_len is 60, len(features)-1000 rows, (60,1000)
      xs.append(features[i-seq_len:i]) #take last 60 days data
      ys.append(targets[i]) #to predict 61th day return
    #       then convert to numpy float32 arrays and return them.
    X=np.array(xs,dtype=np.float32) #conerting python list to nump array, resulting is (no. of seq,no. of days, no. of features)
    y=np.array(ys,dtype=np.float32)
    return X,y

#This function is used to remove the effect of extreme target values (outliers) before training the LSTM.
def winsorise(arr: np.ndarray, n_std: float = 3.0) -> np.ndarray: #arr → A NumPy array (the future returns),n_std → Number of standard deviations to allow (default = 3),It returns a new array where very large or very small values are clipped.
    """
    Clip values in arr that fall outside ±(n_std * arr.std()).
    Applied only to training targets so test evaluation stays unmodified.
    """
    # TODO: compute the symmetric clip bounds and return np.clip(arr, lo, hi).
    std=arr.std()
    lo,hi=-n_std*std,n_std*std #so this sets the range , as std. dev. can be very high as some are 0.01,0.02,0.03,2.5 - so 2.5 sets it to very high
    return np.clip(arr,lo,hi) #np.clip-if value < lo, replace it with value of lo , If value > hi, replace it with value of hi.

split_idx  = int(len(featured) * TRAIN_SPLIT) #no of splits, len(features)-no. of rows, training split= in percent
val_size   = max(int(split_idx * VAL_FRACTION), 1)#val_fraction- in percent
train_end  = split_idx - val_size # total training data=split_idx-val_size, total testing data=total rows-split_idx

scaler = StandardScaler() #some columns have data like 100000, some have 0.1 ,so bring it to same range b/w 0-1
scaler.fit(featured.iloc[:train_end][FEATURE_COLUMNS]) #only training data can compute scaling parameters
all_feats = scaler.transform(featured[FEATURE_COLUMNS]) #now every feature is standardized

raw_targets   = featured["future_return"].to_numpy(dtype=np.float32) # converts the future_return column into a NumPy array:
train_targets = winsorise(raw_targets[:train_end]) # Winsorise only the training targets,takes values till train_end, then winorwise it like 0.01,0.02,1.50,-0.03 become 0.01,0.02,0.12,-0.03
targets       = np.concatenate([train_targets, raw_targets[train_end:]]) #np.concatenate() joins two arrays together.-training targets and remaining targets

x_all, y_all = make_sequences(all_feats, targets, SEQUENCE_LENGTH)
n_train = max(train_end - SEQUENCE_LENGTH, 1) #Although there are 700 training rows, you can't create a sequence for the first 60 rows because there isn't enough history,Although there are 700 training rows, you can't create a sequence for the first 60 rows because there isn't enough history,The first sequence starts at row 60, max(...,1) ensures that there should be atleast 1 training seq. even if dataset is very small
n_val   = max(split_idx - SEQUENCE_LENGTH - n_train, 0)

x_train, y_train = x_all[:n_train], y_all[:n_train]
x_val,   y_val   = x_all[n_train:n_train+n_val], y_all[n_train:n_train+n_val] #val is like mock test
x_test,  y_test  = x_all[n_train+n_val:], y_all[n_train+n_val:]

print(f"train={len(x_train)}  val={len(x_val)}  test={len(x_test)}  "
      f"(sequence length = {SEQUENCE_LENGTH})")
```

## 4. Model: LSTM with temporal attention

A plain LSTM only uses its **last** hidden state, which can miss signals from earlier in the window. **Temporal attention** learns a weighting over all time steps, then combines them into a single context vector.

**Your tasks:**

1. **`TemporalAttention.forward(lstm_out)`** — `lstm_out` has shape `(batch, seq_len, hidden)`.  
   - Apply `self.scorer` (a single Linear layer) to get an unnormalised score per timestep: shape `(batch, seq_len, 1)`.  
   - Normalise across the **time axis** (dim=1) using softmax to get attention weights.  
   - Return the weighted sum of `lstm_out` along the time axis: shape `(batch, hidden)`.

2. **`ReturnSequenceModel.forward(x)`** — `x` has shape `(batch, seq_len, input_size)`.  
   - Pass `x` through `self.recurrent` to get `out` (all hidden states) and `_` (final state, unused).  
   - Pass `out` through `self.attention` to get a single context vector.  
   - Pass the context through `self.head` and squeeze the last dimension to return a 1-D tensor of shape `(batch,)`.


```python
class TemporalAttention(nn.Module): #which days to pay more attention to rather than treating everyday equally
    def __init__(self, hidden_size: int) -> None: # hidden_size= 64 (each lstm output is made up of 64  numbers)
        super().__init__()
        self.scorer = nn.Linear(hidden_size, 1, bias=False) # the scoreer converts each 64 into 1 attention score

    def forward(self, lstm_out: torch.Tensor) -> torch.Tensor: # 32 sequences, 60 days, each day has 64 hidden features
        """
        Args:
            lstm_out: tensor of shape (batch, seq_len, hidden_size)
        Returns:
            context: tensor of shape (batch, hidden_size)
        """
        # TODO: compute per-timestep scores using self.scorer (shape: batch, seq_len, 1)
        # TODO: normalise scores across the time axis (dim=1) via softmax
        # TODO: compute and return the weighted sum of lstm_out along dim=1
        scores=self.scorer(lstm_out) #(32,60,64)->(32,60,1) after linear,each day gets one score
        weights=torch.softmax(scores,dim=1) #softmax converts scores into probabilities, eg- scores(1,2,4)->probabilities(0.04,0.11,0.85), day 3-0.85(so it is important)
        context=torch.sum(weights*lstm_out,dim=1) #without attention-avg of all days, with attention- importnat days contribute more
        return context,weights

#Below code is for neural network- look at last n days of stock and predicts the next day's return , eg-past 60 days of Stock data->learn patterns using LSTM/GRU-> find which days are more imp.(attention)->convert all information into one prediction->tomorrow's return
class ReturnSequenceModel(nn.Module):
    """LSTM / GRU / BiLSTM + temporal attention + small regression head."""
    def __init__(self, input_size, hidden_size=64, num_layers=2,
                 dropout=0.4, model_type="lstm") -> None:
        super().__init__()
        model_type = model_type.lower()
        bidirectional = model_type == "bilstm"
        recurrent_cls = nn.GRU if model_type == "gru" else nn.LSTM
        self.recurrent = recurrent_cls(
            input_size=input_size, hidden_size=hidden_size, num_layers=num_layers,
            dropout=dropout if num_layers > 1 else 0.0,
            batch_first=True, bidirectional=bidirectional,
        )
        out_dim = hidden_size * (2 if bidirectional else 1)
        self.attention = TemporalAttention(out_dim)
        self.head = nn.Sequential(
            nn.LayerNorm(out_dim),
            nn.Linear(out_dim, 64),
            nn.GELU(),
            nn.Dropout(dropout * 0.5),
            nn.Linear(64, 1),
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor: # forward() function- describes how data flows through your neural network. It is executed every time you pass an input to the model.
        """
        Args:
            x: tensor of shape (batch, seq_len, input_size)
        Returns:
            predictions: tensor of shape (batch,)
        """
        # TODO: pass x through self.recurrent — unpack the tuple to get `out` (all hidden states)
        # TODO: pass out through self.attention to get a context vector
        # TODO: pass context through self.head and squeeze the last dim
        rnn_out, _=self.recurrent(x) # input-(32,60,18)- 32 stock seq. ,each seq. has 60 days, each day has 18 features-this input goes into LSTM/GRU - the lstm returns 64 numbers for each day as a output
        context, _=self.attention(rnn_out) # output of the LSTM goes into the attention layer.
        out=self.head(context) #now context goes into pred. head -(32,64)->(32,1)
        return out.squeeze(-1) #removes dim. of size 1- (32,1)->(32)
```

## 5. Loss function: direction-aware, class-balanced

Plain regression loss (MSE/Huber) has no notion of *sign*. This custom loss adds an explicit direction signal.

**Your task — implement `DirectionWeightedLoss.forward(pred, target)`:**

The loss has two components that are summed:

1. **Weighted Huber term** — Use `self.huber(pred, target)` (returns per-sample losses, shape `(batch,)`).  
   Build a per-sample weight tensor `w`: `self.up_weight` where `target > 0`, else `self.down_weight`.  
   Return the mean of `huber_losses * w`.

2. **Weighted BCE term** — Convert the regression predictions to a direction signal by multiplying by `self.temp` (a temperature that sharpens the soft-sigmoid).  
   Compute `binary_cross_entropy_with_logits` between `pred * self.temp` and `(target > 0).float()`, with `reduction="none"`.  
   Take the mean of `bce_losses * w` (same weights as above) and scale by `self.lambda_dir`.

Return the sum of both terms.

`compute_class_weights` is already implemented — it returns `(up_weight, down_weight)` as balanced inverse-frequency weights.


```python
class DirectionWeightedLoss(nn.Module): #creates a loss funcn
    def __init__(self, up_weight=1.0, down_weight=1.0, lambda_dir=0.30, temp=50.0): #Without weights, the model can not be trained on down days(80% up days and 20% down days)
        super().__init__()
        self.up_weight, self.down_weight = up_weight, down_weight
        self.lambda_dir, self.temp = lambda_dir, temp
        self.huber = nn.SmoothL1Loss(reduction="none")

    def forward(self, pred: torch.Tensor, target: torch.Tensor) -> torch.Tensor: #predicted return vs actual return, past 60 days->neural network->predicted return -> compare with actual return -> calculate loss -> Backpropagation-> update model weights -> better pred.
        """
        Args:
            pred:   predicted returns, shape (batch,)
            target: actual returns,    shape (batch,)
        Returns:
            scalar loss combining weighted Huber regression and weighted BCE direction terms
        """
        # TODO: build per-sample weight tensor w (up_weight where target > 0, else down_weight)
        # TODO: compute weighted Huber term: mean of self.huber(pred, target) * w
        # TODO: compute dir_target = (target > 0).float()
        # TODO: compute per-sample BCE using binary_cross_entropy_with_logits(pred * self.temp, dir_target, reduction="none")
        # TODO: compute weighted BCE term: mean of bce_per_sample * w, scaled by self.lambda_dir
        # TODO: return the sum of both terms
        w= torch.where(target>0, torch.full_like(target,self.up_weight),torch.full_like(target,self.down_weight)) # this gives weight to every sample, target>0 -if true use up_weight, if false use down_weight, 1- stock went up , 0- stock went down.
        huber_term=(self.huber(pred,target)*w).mean() #diff. b/w pred. and target and then mutltiply with weight and then take avg.
        dir_target=(target>0).float() # is tomorrow up or down? , up-1 and down-0
        bce_per_sample=torch.nn.functional.binary_cross_entropy_with_logits(pred*self.temp,dir_target,reduction='none') #now the model is checked on dirn like if pred-up and actual-up , then ok otherwise if contrary then huge bce loss , if pred=0.02 then multiply with self.temp=50 to make it 1
        bce_term=(bce_per_sample*w).mean()*self.lambda_dir # rare classes get higher weights, then multiply with lambda_dir=0.3
        return huber_term + bce_term

def compute_class_weights(y_train: np.ndarray): #this funcn canlculates the weights automatically
    """Balanced weights: N / (2 * n_class) for each direction."""
    n      = len(y_train) #total training samples
    n_up   = max((y_train > 0).sum(), 1) #up samples
    n_down = max((y_train <= 0).sum(), 1) # down samples
    return float(n / (2.0 * n_up)), float(n / (2.0 * n_down)) #calculate weight of up and down sample.
```

## 6. Training loop

Key design choices in the training loop below:
- **`ReduceLROnPlateau`** drops the learning rate only when validation loss genuinely stalls.
- **Early stopping** on validation loss restores the best-loss checkpoint at the end.

**Your task — inside `train_model()`, complete the per-batch training step:**

Each iteration over `loader` yields `(bx, by)` (a batch of input sequences and targets). You must:
1. Move `bx` and `by` to `DEVICE`.
2. Zero the optimizer gradients.
3. Compute the model's output and the batch loss using `loss_fn`.
4. Run backpropagation.
5. Clip gradients to `config.grad_clip` using `nn.utils.clip_grad_norm_`.
6. Take an optimizer step.

The optimizer construction (`AdamW`), scheduler, validation loop, early-stopping logic, and metric collection are all already written for you.


```python
@dataclass
class TrainConfig:
    epochs: int = 5
    batch_size: int = 64
    learning_rate: float = 0.05
    weight_decay: float = 1e-3
    grad_clip: float = 1.0
    patience: int = 10
    lambda_dir: float = 0.30
    hidden_size: int = 8
    num_layers: int = 2
    dropout: float = 0.4
    model_type: str = "lstm"


def train_model(x_train, y_train, x_val, y_val, x_test, y_test,
                 config: TrainConfig, verbose=True):
    w_up, w_down = compute_class_weights(y_train)
    loss_fn = DirectionWeightedLoss(up_weight=w_up, down_weight=w_down,
                                     lambda_dir=config.lambda_dir)

    model = ReturnSequenceModel(
        input_size=x_train.shape[-1], hidden_size=config.hidden_size,
        num_layers=config.num_layers, dropout=config.dropout,
        model_type=config.model_type,
    ).to(DEVICE)

    loader = DataLoader(TensorDataset(torch.tensor(x_train), torch.tensor(y_train)),
                         batch_size=config.batch_size, shuffle=True)
    optimizer = torch.optim.AdamW(model.parameters(), lr=config.learning_rate,
                                   weight_decay=config.weight_decay)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="min", factor=0.5, patience=5, min_lr=config.learning_rate * 0.02)

    history = {"train_loss": [], "val_loss": [], "val_dir_acc": [], "lr": []}
    best_val, best_state, patience_ctr = float("inf"), None, 0
    best_acc, best_acc_epoch = -1.0, -1

    for epoch in range(config.epochs):
        model.train()
        ep_losses = [] #creates an empty list, store the loss of every time and then the loop runs loss is improved and finally avg. is calculated
        for bx, by in loader:
            # TODO: move bx and by to DEVICE
            bx,by=bx.to(DEVICE),by.to(DEVICE) #bx-batch of input seq(64,60,18), by-actual future returns(64,)
            # TODO: zero the optimizer gradients
            optimizer.zero_grad() # clears gradent for every iteration otherwise it accumulates.
            # TODO: forward pass: compute loss_fn(model(bx), by)
            preds=model(bx) #forward pass, past 60 days of input->lstm->attention->prediction
            loss=loss_fn(preds,by) #now compare pred vs actual, how wrong our model is
            # TODO: backward pass
            loss.backward() #backpropagation- which weights causes this error?, pytorch computes gradient for every parameter , this gradient tells the optimzer how to change?
            # TODO: clip gradients: nn.utils.clip_grad_norm_(model.parameters(), config.grad_clip)
            nn.utils.clip_grad_norm_(model.parameters(),config.grad_clip) #limts higher data like 1200 is adjusted to , otherwise very much voltaile the model will become
            # TODO: optimizer step
            optimizer.step() #now the optimizer updates the model weight, makes the model slighhtly better with every iteration
            # TODO: append loss.item() to ep_losses
            ep_losses.append(loss.item()) #loss is added
        train_loss = float(np.mean(ep_losses))

        model.eval()
        with torch.no_grad():
            vp = model(torch.tensor(x_val).to(DEVICE))
            vy = torch.tensor(y_val).to(DEVICE)
            val_loss  = float(loss_fn(vp, vy).item())
            val_preds = vp.cpu().numpy()
            val_acc   = float(accuracy_score(y_val > 0, val_preds > 0))
        scheduler.step(val_loss)

        if val_loss < best_val - 1e-6:
            best_val, best_state, patience_ctr = val_loss, \
                {k: v.clone() for k, v in model.state_dict().items()}, 0
        else:
            patience_ctr += 1
        if val_acc > best_acc:
            best_acc, best_acc_epoch = val_acc, epoch + 1
        if patience_ctr >= config.patience:
            if verbose:
                print(f"  Early stop @ epoch {epoch+1} (best val={best_val:.6f})")
            break

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["val_dir_acc"].append(val_acc)
        history["lr"].append(optimizer.param_groups[0]["lr"])

        if verbose and ((epoch + 1) % 10 == 0 or epoch == 0):
            print(f"  Epoch {epoch+1:03d}  train={train_loss:.6f}  val={val_loss:.6f}  "
                  f"dir_acc={val_acc:.3f}  lr={optimizer.param_groups[0]['lr']:.2e}")

    if best_state:
        model.load_state_dict(best_state)
    if verbose and best_acc_epoch > 0 and best_acc_epoch < len(history["val_loss"]) - 3:
        print(f"  [note] best val_dir_acc={best_acc:.3f} was at epoch {best_acc_epoch}, "
              f"but the deployed checkpoint is the lowest-val-loss one instead.")

    model.eval()
    with torch.no_grad():
        test_preds = model(torch.tensor(x_test).to(DEVICE)).cpu().numpy()

    metrics = {
        "test_rmse":          float(np.sqrt(mean_squared_error(y_test, test_preds))),
        "test_mae":           float(mean_absolute_error(y_test, test_preds)),
        "direction_accuracy": float(accuracy_score(y_test > 0, test_preds > 0)),
        "f1_score":           float(f1_score(y_test > 0, test_preds > 0, zero_division=0)),
        "n_train": len(x_train), "n_val": len(x_val), "n_test": len(x_test),
        "y_test": y_test, "preds": test_preds,
    }
    return model, history, metrics
```

```python
config = TrainConfig()

baseline_model, baseline_history, baseline_metrics = train_model(
    x_train, y_train, x_val, y_val, x_test, y_test, config
)

print("\nBaseline test metrics:")
for k, v in baseline_metrics.items():
    if k not in ("y_test", "preds"):
        print(f"  {k}: {v}")
```

## 7. Actual vs predicted price

The model predicts next-day **return**, not price directly. To plot it against the actual price
series, we reconstruct an implied price path: `predicted_price[t+1] = actual_price[t] *
(1 + predicted_return[t+1])`. This is a one-step-ahead reconstruction — each predicted point uses
the **true** price from the day before, not a chained prediction, so errors don't compound across
the chart. It's the right way to visually sanity-check a return-prediction model against price.


```python
test_start_idx = len(featured) - len(y_test)
test_dates  = featured["date"].iloc[test_start_idx:].reset_index(drop=True)
actual_price_at_t = featured["close"].iloc[test_start_idx - 1 : -1].reset_index(drop=True)

actual_next_price    = actual_price_at_t.values * (1 + baseline_metrics["y_test"])
predicted_next_price = actual_price_at_t.values * (1 + baseline_metrics["preds"])

plt.figure(figsize=(14, 5))
plt.plot(test_dates, actual_next_price, label="Actual price", color="#2ecc71", linewidth=1.3)
plt.plot(test_dates, predicted_next_price, label="Predicted price", color="#e74c3c",
         linewidth=1.1, alpha=0.85)
plt.title("Baseline LSTM — Actual vs Predicted Price (test set)")
plt.xlabel("Date"); plt.ylabel("Price ($)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(f"{REPORTS_DIR}/baseline_actual_vs_predicted_price.png", dpi=140)
plt.show()
```
